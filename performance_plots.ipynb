```python
import json
def load_json(p: str):
    with open(p) as f:
        return json.load(f)
```

```python
vec_load_json = np.vectorize(load_json)

baseline_json = vec_load_json(baseline_result_files)
unstable_communication_json = vec_load_json(unstable_communication_result_files)
response_misreporting_json = vec_load_json(response_misreporting_results_files)
```

```python
def collaboration_rate(domain_values: np.ndarray):
    return 'Collaboration Rate', (domain_values[..., 1] / (domain_values[..., 0] * 1.0)).mean(axis=0)

def own_coin_rate(domain_values: np.ndarray):
    return 'P(Own Coin)', np.divide(domain_values[..., 2], domain_values[..., 1], out=np.zeros_like(domain_values[..., 2]), where=domain_values[..., 1]!=0).mean(axis=0)

def sustainability(domain_values: np.ndarray):
    return 'Sustainability', domain_values[..., 3].mean(axis=0) / (domain_values[..., 0].mean(axis=0) * 1.0)
```

```python
unstable_communication_labels = [f'DRIVE Comm. Failure ({round(perc, 1)})' for perc in np.arange(0.1, 1.1, 0.1)]
response_misreporting_labels = [f'DRIVE Response Misreporting ({round(perc, 1)} - 0.9)' for perc in np.arange(0.1, 1.1, 0.1)]
```

```python
from datetime import datetime
from matplotlib import pyplot as plt

game_names_unstable_communication = ['Coin-2', 'Coin-4']
game_names_response_misreporting = ['Coin-2', 'Coin-4']
domain_value_metric_functions_unstable_communication = [own_coin_rate, own_coin_rate]
domain_value_metric_functions_response_misreporting = [own_coin_rate, own_coin_rate]

for failure_type, labels, json, game_names, domain_value_metric_functions in zip(
        ['unstable_communication', 'response_misreporting'],
        [unstable_communication_labels, response_misreporting_labels],
        [unstable_communication_json, response_misreporting_json],
        [game_names_unstable_communication, game_names_response_misreporting],
        [domain_value_metric_functions_unstable_communication, domain_value_metric_functions_response_misreporting],
):
    algo_names = ['DRIVE'] + labels

    fig, axs = plt.subplots(json.shape[0], 1, figsize=[12, json.shape[0] * 8])
    fig.suptitle(failure_type.replace('_', ' ').capitalize())

    cmap = plt.get_cmap('jet')

    axs = np.array([axs]) if type(axs) is not np.ndarray else axs # if only one subplot

    for i, (game_name, data, fun) in enumerate(zip(game_names, json, domain_value_metric_functions)):
        ax = axs[i]
        for j, (algo_name, d) in enumerate(zip(algo_names, np.concatenate((baseline_json[i], data)))):
            domain_values = np.array([x['domain_values'] for x in d])
            metric_name, mean_domain_values = fun(domain_values)

            ax.plot(list(range(len(mean_domain_values))), mean_domain_values, alpha=0.5, label=algo_name, color=cmap(j/len(algo_names)))
            ax.set_title(f'{game_name} - {domain_values.shape[0]} runs')
            ax.set_ylabel(metric_name)
            ax.set_xlabel('Epochs')
            ax.legend()
        ax.grid(alpha=0.3)
    fig.tight_layout()
    import os

    os.makedirs(f"plots/{failure_type}", exist_ok=True)
    plt.savefig(f'plots/{failure_type}/performance_summary_{failure_type}_{datetime.now().strftime("%Y-%m-%d_%H-%M-%S")}.png')
```
